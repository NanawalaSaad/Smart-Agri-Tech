```python
import pandas as pd
import pickle
import os

print("--- DESERT AGRITECH: FEATURE ENGINEERING ---")

# 1. Naya 250 crops wala dataset load karo
df = pd.read_csv('../data/processed/desert_crops_250.csv')
print(f"Loaded Shape: {df.shape}")

# 2. Priority Scaling (Normalization)
# Taki algorithm ko calculate karne mein aasaani ho (0 to 1 scale)
df['priority_weight'] = df['priority'] / 10.0

# 3. Environment Stringency Index
# Ye batayega ki crop kitni "tough" hai (High temp + Low water)
df['toughness_index'] = (df['temp_max'] / 50) + (1 - (df['rain_max'] / 4000))
df['toughness_index'] = df['toughness_index'].round(2)

print("\nFeatures Created:")
print(df[['name', 'priority_weight', 'toughness_index']].head())
```

```python
# Create models directory if not exists
os.makedirs('../models', exist_ok=True)

# Convert DataFrame to a List of Dictionaries (Best for fast API searching)
crops_database = df.to_dict(orient='records')

# Save as a .pkl file
db_path = '../models/expert_system_db.pkl'
with open(db_path, 'wb') as f:
    pickle.dump(crops_database, f)

print(f"\n✅ Expert System Database saved successfully at: {db_path}")
print(f"Total Crops in DB: {len(crops_database)}")
```
