```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt 
import seaborn as sns 

#dataset load karo
df=pd.read_csv('../data/raw/desert_hydroponic_dataset.csv')

print("Shape:",df.shape)
print("Columns:",df.columns.tolist())
df.head()
```

```python
#Data Cleaning Check
#Null Values check karo
print("----NULL VALUES----")
print(df.isnull().sum())

#DUPLICATE ROWS CHECK KARO
print("\n----DUPLICATE ROWS----")
print("Duplicates:",df.duplicated().sum())

#OUTLIERS check karo
print("----MIN/MAX CHECK---")
print(df.describe().round(2))
```

```python
#FEATURE ENGINEERING
#naye smart columns banayenge:

#1.Heat Stress index
#temperature + humidity combine karke ek index banayenge
df['heat_stress_index']=df['temperature']*(1-df['humidity']/100)

#2.Water Efficiency
#kitna paani diya vs soil moisture kitna hai
df['water_efficiency']=df['water_given_ml']/(df['soil_moisture']+1)

#3. pH Deviation
#ideal pH 6.5 se kitna door hai
df['pH_deviation']=abs(df['pH_level']-6.5)

#4. Growth Score
#light+nutrients combined score
df['growth_score']=(df['light_hours']*df['nutrient_level'])/100


#naye columns dekho
print("Naye Features:")
print(df[['heat_stress_index','water_efficiency','pH_deviation','growth_score']].head())
print("\nNew shape:",df.shape)
```

```python
#clean + engineered data save karo
df.to_csv('../data/processed/desert_hydroponic_processed.csv',index=False)

print("Processed Data Saved!")
print("location:data/processed")
print("shape:",df.shape)
```
