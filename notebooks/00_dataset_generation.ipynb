```python
import pandas as pd 
import numpy as np 

#seed
np.random.seed(42)

print("Libraries Imported!")
```

```python
#Dataset Size
#rows
n_samples=5000
print(f"IN Dataset {n_samples} rows there")
```

```python
temperature=np.random.uniform(35,55,n_samples) # Desert temp
humidity=np.random.uniform(10,40,n_samples) #Dry Desert
soil_moisture=np.random.uniform(20,80,n_samples) #Soil
pH_level=np.random.uniform(5.5,7.5,n_samples) #pH
light_hours=np.random.uniform(6,14,n_samples) #SunLight
water_given_ml=np.random.uniform(100,500,n_samples) #water
nutrient_level=np.random.uniform(0,100,n_samples) #Nutrients
co2_ppm=np.random.uniform(400,1200,n_samples) #CO2
day_number=np.random.uniform(1,91,n_samples) #Crop cycle day

print("Features are ready!")
```

```python
#Target(Yield) now
yield_kg=(
    (water_given_ml*0.008)+ #zyada pani=zyada yield[more water = more yield]
    (nutrient_level*0.05)+ #nutrients important
    (light_hours*0.3)+ #sunlight
    (soil_moisture*0.02)- #moisture help karta hai
    (temperature*0.1)- #zyada heat=nuksan
    (np.abs(pH_level-6.5)*0.5)+  #6.5 ideal pH
    np.random.normal(0,0.5,n_samples) #real world noise
)

#negative yield nahi hoti
yield_kg=np.clip(yield_kg,0.1,15)
print("Yield calculated!")
print(f"Min Yield: {yield_kg.min():.2f} kg")
print(f"Max Yield: {yield_kg.max():.2f} kg")
```

```python
#plant Stressed hai ya nai
water_stress=np.where(
    (soil_moisture<30) | (temperature>45),
    1, #stressed
    0 #healthy 
)

stress_percent=water_stress.sum() / n_samples *100
print(f"Stressed plants:{stress_percent:.1f}%")
```

```python
df=pd.DataFrame({
    'temperature':temperature,
    'humidity':humidity,
    'soil_moisture':soil_moisture,
    'pH_level':pH_level,
    'light_hours':light_hours,
    'water_given_ml':water_given_ml,
    'nutrient_level':nutrient_level,
    'co2_ppm':co2_ppm,
    'day_number':day_number,
    'yield_kg':yield_kg,
    'water_stress':water_stress
})

#save karo
df.to_csv('../data/raw/desert_hydroponic_dataset.csv',index=False)

print("Dataset Saved!")
print(f"Shape:{df.shape}")
df.head()
```
