```python
# Pehle apne teeno lists ko ek Dictionary (Database) mein combine karna padega
crops_db = {
    'trees': trees,          # Tere 100 trees
    'plants': plants,        # Tere 100 plants
    'dry_fruits': dry_fruits # Tere 50 dry fruits
}

def recommend_crops(sensor_data, top_n=3):
    results = {'trees': [], 'plants': [], 'dry_fruits': []}

    for category, crops in crops_db.items():
        scores = []
        for crop in crops:
            score = 0
            total = 0

            # Temperature match (25%)
            if crop['temp_min'] <= sensor_data['temp'] <= crop['temp_max']:
                score += 25
            elif abs(sensor_data['temp'] - crop['temp_min']) <= 5 or \
                 abs(sensor_data['temp'] - crop['temp_max']) <= 5:
                score += 12
            total += 25

            # Humidity match (20%)
            if crop['humidity_min'] <= sensor_data['humidity'] <= crop['humidity_max']:
                score += 20
            elif abs(sensor_data['humidity'] - crop['humidity_min']) <= 10 or \
                 abs(sensor_data['humidity'] - crop['humidity_max']) <= 10:
                score += 10
            total += 20

            # Soil moisture match (20%)
            if crop['soil_moisture_min'] <= sensor_data['soil_moisture'] <= crop['soil_moisture_max']:
                score += 20
            elif abs(sensor_data['soil_moisture'] - crop['soil_moisture_min']) <= 10 or \
                 abs(sensor_data['soil_moisture'] - crop['soil_moisture_max']) <= 10:
                score += 10
            total += 20

            # pH match (15%)
            if crop['ph_min'] <= sensor_data['ph'] <= crop['ph_max']:
                score += 15
            elif abs(sensor_data['ph'] - crop['ph_min']) <= 0.5 or \
                 abs(sensor_data['ph'] - crop['ph_max']) <= 0.5:
                score += 7
            total += 15

            # Light match (10%)
            if crop['light_min'] <= sensor_data['light'] <= crop['light_max']:
                score += 10
            total += 10

            # Rain match (10%) - Defaulting to Annual 500mm for calculations
            if crop['rain_min'] <= sensor_data.get('rain_annual_mm', 500) <= crop['rain_max']:
                score += 10
            total += 10

            # Priority bonus (Saudi/India gets extra marks)
            priority_bonus = crop['priority'] * 0.5
            score += priority_bonus

            # Final percentage calculation
            pct = round((score / (total + 5)) * 100, 1)
            pct = min(pct, 99.9)  # Cap at 99.9%

            scores.append({
                'name':       crop['name'],
                'region':     crop['region'],
                'match_pct':  pct,
                'grow_days':  crop['grow_days'],
                'water_need': crop['water_need'], # Removed yield_kg to fix KeyError
            })

        # Sort by match %
        scores.sort(key=lambda x: x['match_pct'], reverse=True)
        results[category] = scores[:top_n]

    return results

# Test with Live Sensor Data (Example: Your ESP32 values)
test_data = {
    'temp': 29.7,          # Tera OLED wala temp
    'humidity': 85.2,      # Teri ESP32 Humidity
    'soil_moisture': 100,  # Tera wet soil
    'ph': 2.56,            # Tera pH sensor (Danger low pH)
    'light': 57,           # Tera LDR %
    'rain_annual_mm': 600  # Ek average annual rainfall set kar diya
}

recommendations = recommend_crops(test_data)

print("="*55)
print("  🌾 Desert AgriTech — Crop Recommendations 🌾")
print("="*55)
print(f"  Live Sensor: Temp={test_data['temp']}°C | "
      f"Humidity={test_data['humidity']}% | "
      f"Soil={test_data['soil_moisture']}% | "
      f"pH={test_data['ph']}")
print("="*55)

for category, crops in recommendations.items():
    emoji = '🌳' if category=='trees' else '🌿' if category=='plants' else '🥜'
    print(f"\n{emoji} {category.upper()}:")
    for i, crop in enumerate(crops, 1):
        print(f"  {i}. {crop['name']:20s} | Match: {crop['match_pct']:5.1f}% | "
              f"Grow: {crop['grow_days']} days | "
              f"Water: {crop['water_need']}")
```
